# 16 — ParaView visualization notes

This is a lightweight reference notebook for viewing exported `.vti` files in ParaView.

## Open the VTI

```text
File → Open → sims_volume.vti
Apply
```

In the top toolbar, choose the scalar channel:

```text
Color By → Total / Si / Ti / Mg / ...
```

## Fast layer scan

Use a `Slice` filter.

```text
Filters → Common → Slice
Normal = (0, 0, 1)
```

Then change:

```text
Origin → Z
```

to move layer-by-layer through the SIMS stack.

## Cut away a corner

Use a box clip.

```text
Filters → Common → Clip
Clip Type → Box
Invert → ON
```

Shrink the box and move it into one corner. With `Invert` on, ParaView removes the inside of the box,
giving a cutaway corner view.

## Show interfaces as surfaces

For clear interfaces, use surfaces instead of volume fog:

```text
Dataset → Contour → Surface → Clip
```

Workflow:

```text
1. Select dataset
2. Filters → Common → Contour
3. Select scalar, e.g. Si
4. Choose isovalue threshold
5. Representation → Surface
6. Filters → Clip → Box → Invert
```

This gives solid buried interface surfaces.

## Recommended SIMS tomography visualization

Use two or more pipeline objects:

```text
Total counts → Volume rendering, gray, low opacity
Element 1    → Contour surface
Element 2    → Contour surface
Element 3    → Contour surface
```

Then apply the same clipping box to expose the internal interface.
